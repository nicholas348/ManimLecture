```python
from manim import *
```

```python
%%manim -qm -p  my_first_manim_animation.py MyFirstManimAnimation
class MyFirstManimAnimation(Scene):
    def construct(self):

        # 等待一秒
        self.wait(1)
        hello_manim = (Tex("hello, Manim!")#创造一个Manim文字对象
                       .scale(3))#放大3倍

        # 书写 hello world
        self.add(hello_manim)
```
