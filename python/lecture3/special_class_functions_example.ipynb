# 特殊方法

```python
class Vector:
    def __init__(self, x, y):
        """这个方法用于初始化vector"""
        self.x = x
        self.y = y

    def __repr__(self):
        """在print是输出什么"""
        return f"Vector({self.x}, {self.y})"

    def __str__(self):
        """在打印时输出什么"""
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        """在执行 + 运算时返回什么"""
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    def __sub__(self, other):
        """在执行 - 运算是返回什么"""
        return Vector(self.x - other.x, self.y - other.y)

    def __eq__(self, other):
        """判断是否相等的条件（==)"""
        return self.x == other.x and self.y == other.y
    def __len__(self):
        """输出长度"""
        # For a vector, let's return the number of dimensions
        return 2

    def __call__(self, scalar):
        """使得这个物体可以像函数一样被引用"""
        return Vector(self.x * scalar, self.y * scalar)
    def __mul__(self, other):
        return Vector(self.x*other, self.y*other)
```

## 使用

创建一个vector类对象 v1，初始值为（2，4）

```python
v1 = Vector(2, 4)
```

创建另一个vector类对象v2，初始值为（1，3）

```python
v2 = Vector(1, 3)
```

打印v1,使用v1的`__repr__`方法

```python
print(f"v1: {v1}")               #  (2, 4)
```

打印v1与v2的和，使用`__add__`方法

```python
print(f"Sum: {v1 + v2}")         # (3, 7)
```

打印v1是否等于v2，使用`__eq__`方法

```python
print(f"Equal? {v1 == v2}")      #  False
```

打印v1作用在10上的结果，使用`__call__`方法

```python
print(f"Scaled: {v1(10)}")       # (20 , 40)
```

打印v1拉伸10倍，使用`__mul__`方法

```python
print(f"Scaled: {v1*10}")       #  (20 , 40)
```

# 保护方法



```python
class MyClass:

    public_data = "任何人都可以看见这个"
    _internal_data = "私密数据"
    __more_internal_data = "极度私密数据（密码等）"

    """
    无下划线：普通方法
    """
    @classmethod
    def normal_method(cls):
        return cls.public_data

    """
    开发者不想让你引用的方法（可能随时都会改名或直接删掉）
    不建议引用
    """

    """
    单个下划线：保护方法
    """
    @classmethod
    def _internal_method(cls):
        return cls._internal_data

    """
    二重下划线：加重保护
    """
    @classmethod
    def __more_internal_method(cls):
        return cls.__more_internal_data
```

调用普通方法

```python
print(MyClass.normal_method())
```

调用保护方法

```python
print(MyClass._internal_method())
```

调用加重保护方法

```python
print(MyClass._MyClass__more_internal_method())
```
