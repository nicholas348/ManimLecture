# 常见的类装饰器

```python
class Circle:
    def __init__(self, radius: float):
        self._radius = radius

    # 1. @property - Turns a method into a read-only attribute
    @property
    def radius(self) -> float:
        return self._radius

    # 2. @property.setter - 在给类参数赋值时可以增加判断条件,如果不符条件可以报错
    @radius.setter
    def radius(self, value: float):
        if value < 0:
            raise ValueError("Radius cannot be negative")
        self._radius = value

    # 3. @staticmethod -属于这个类的函数，不关联这个class的存储的数值
    @staticmethod
    def is_valid_radius(value: float) -> bool:
        return value > 0

    # 4. @classmethod - 仅需接受在class内部存储的变量，无需关系到实例
    @classmethod
    def from_diameter(cls, diameter: float):
        return cls(diameter / 2)
```

# 使用特例

设置特征

```python
c = Circle(5)
print(f"Initial Radius: {c.radius}")
```

设置一个合法的直径

```python
c.radius = 10  # Uses the setter
```

设置一个不合法的半径

```python
c.radius=-3
```

检查直径

```python
print(f" Radius: {c.radius}")
```

输出circle中，半径为-5是否合法

```python
print(f"Is -5 valid? {Circle.is_valid_radius(-5)}")
```

class的方法

```python
c2 = Circle.from_diameter(20)
print(f"Radius from diameter 20: {c2.radius}")
```
