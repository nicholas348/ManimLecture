```python
first_string = "Hello"
second_string = 'World'
```

int转str

```python
str_from_int = str(123) #结果: '123'
```

float转str

```python
str_from_float = str(45.67) #结果: '45.67'
```

string转int

```python
int_from_string = int("89")  #结果: 89
```

string转float

## 字符串的方法

拼接

```python
string_concat = first_string + " " + second_string #结果: 'Hello World'
```

重复

```python
string_repeat = first_string * 3 #结果: 'HelloHelloHello'
```

长度

```python
string_len = len(first_string) #结果: 5
```

转换为大写

```python
string_upper = first_string.upper() #结果: 'HELLO'
```

转换为小写

```python
string_lower = string_upper.lower() #结果: 'hello'
```

```python
print(float("56.78abc")) # 错误：ValueError: could not convert string to float: '56.78abc'
```

```python
print(float(""))        # 错误：ValueError: could not convert string to float: ''
```

# 4. 布尔值 (bool)

## 布尔值定义与类型转换

```python
first_bool = True
second_bool = False
```

非0数字转True

```python
bool_from_one = bool(1) #结果: True
```

0转False

```python
bool_from_zero = bool(0)# 结果: False
```

负数转True

```python
bool_from_neg = bool(-1)#结果: True
```

空字符串转False

```python
bool_from_empty_str = bool("")  #结果: False
```

非空字符串转True

```python
bool_from_non_empty_str = bool("False")  # 结果: True
```

bool转int, False=0, True=1

```python
int_from_bool = int(False)  # 结果: 0
```

## 布尔值的逻辑方法

逻辑与

```python
bool_and = first_bool and second_bool  #  结果: False
```

逻辑或

```python
bool_or = first_bool or second_bool  #结果: True
```

逻辑非

```python
bool_not = not first_bool  # 结果: False
```

比较

```python
bool_equal = (1 == 1) # 结果: True
```

## 布尔值的常见错误

 (bool常作为条件使用，直接的操作错误较少)

```python
bool_add = True + False  #不常见，但可执行，True=1, False=0 (结果: 1)
```

```python
print(bool("abc") * 2) #可执行，True=1 (结果: 2)
```
