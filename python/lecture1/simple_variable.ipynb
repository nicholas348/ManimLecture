整数的算术方法

```python
number_one = 25
number_two = 4
```

加法

```python
number_add = number_one + number_two #结果: 29
```

减法

```python
number_sub = number_one - number_two  #结果: 21
```

乘法

```python
number_mul = number_one * number_two #结果: 100
```

除法，结果为浮点数

```python
number_div = number_one / number_two #结果: 6.25
```

整除，结果为整数

```python
number_floordiv = number_one // number_two #结果: 6
```

取模/余数

```python
number_mod = number_one % number_two #结果: 1
```

乘方

```python
number_pow = number_one ** number_two #结果: 390625
number_pow
```

```python
number_abs = abs(-5)
```
