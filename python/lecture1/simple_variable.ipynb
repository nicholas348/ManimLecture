# 2. 浮点数 (float)

## 浮点数定义与类型转换

```python
first_float = 1.5
second_float = 0.5
```

科学计数法

```python
float_scientific = 3e10  # 结果: 30000000000.0
```

int转float

```python
float_from_int = float(3) #结果: 3.0
```

float转int, 截断小数部分

```python
int_from_float = int(2.47) # 结果: 2
```

负float转int, 截断小数部分

```python
int_from_neg_float = int(-2.54) #结果: -2
```

## 浮点数的算术方法

加法

```python
float_add = first_float + second_float #结果: 2.0
```

减法

```python
float_sub = first_float - second_float #结果: 1.0
```

乘法

```python
float_mul = first_float * second_float #结果: 0.75
```

除法

```python
float_div = first_float / second_float #结果: 3.0
```

整除，结果仍为浮点数

```python
float_floordiv = first_float // second_float #结果: 3.0
```

取模

```python
float_mod = first_float % second_float  # 结果: 0.0
```

乘方

```python
float_pow = first_float ** second_float  # 结果: 1.224744871391589(近似舍位数结果）
```

```python
float_is_integer = (3.0).is_integer() #结果: True
```

```python
print(5.0 / 0)          # 0不能作为除数
```

# 4. 布尔值 (bool)

## 布尔值定义与类型转换

```python
first_bool = True
second_bool = False
```

非0数字转True

```python
bool_from_one = bool(1) #结果: True
```

0转False

```python
bool_from_zero = bool(0)# 结果: False
```

负数转True

```python
bool_from_neg = bool(-1)#结果: True
```

空字符串转False

```python
bool_from_empty_str = bool("")  #结果: False
```

非空字符串转True

```python
bool_from_non_empty_str = bool("False")  # 结果: True
```

bool转int, False=0, True=1

```python
int_from_bool = int(False)  # 结果: 0
```

bool转string

```python
string_from_bool_true = str(True)#结果：'true'
string_from_bool_true
```

```python
string_from_bool_false = str(False) #结果：'false'
```

## 布尔值的逻辑方法

逻辑与

```python
bool_and = first_bool and second_bool  #  结果: False
```

逻辑或

```python
bool_or = first_bool or second_bool  #结果: True
```

逻辑非

```python
bool_not = not first_bool  # 结果: False
```

比较

```python
bool_equal = (1 == 1) # 结果: True
```

## 布尔值的常见错误

 (bool常作为条件使用，直接的操作错误较少)

```python
bool_add = True + False  #不常见，但可执行，True=1, False=0 (结果: 1)
```

```python
print(bool("abc") * 2) #可执行，True=1 (结果: 2)
```

# 5. 复数 (complex)

定义：形如 $a+bi$的数，其中$i^2=-1$

##复数的定义

```python
first_complex = 3 + 4j
```

使用两个实数对创建复数

```python
second_complex = complex(1, 2)#结果: 1+2j
```

int转复数

```python
complex_from_int = complex(1)#结果：(1+0j)
```

float转复数

```python
complex_from_float = complex(1.2543)#结果：(1.2543+0j)
```

string 转复数

```python
complex_from_string = complex("1+2j")#结果：(1+2j)
```

bool转复数

```python
complex_from_false = complex(False)#结果：(0j)
```

```python
complex_from_true = complex(True)#结果：(1+0j)
```

## 复数的方法

加法

```python
complex_add = first_complex + second_complex #结果: 4+6j
```

减法

```python
complex_sub = first_complex-second_complex #结果：2+2j
```

乘法

```python
complex_mul = first_complex * second_complex #结果: -5+10j
```

除法

```python
complex_div = first_complex / second_complex #结果：2.2-0.4j
```

实部

```python
complex_real = first_complex.real #结果: 3.0
```

虚部

```python
complex_imag = first_complex.imag #结果: 4.0
```

共轭复数

```python
complex_conj = first_complex.conjugate() #结果: 3-4j
```

复数的常见错误

```python
print(3 + j)             # 错误：NameError: name 'j' is not defined (虚部必须有数字)
```

```python
print(complex("abc"))  # 错误：ValueError: complex() arg is a malformed string
```

```python
print(complex("1+2i")) # 错误：ValueError: complex() arg is a malformed string (必须用j表示虚部)
```
