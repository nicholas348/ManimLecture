# 1.列表（List）- 有序、可变、支持混合类型元素

## 列表定义与类型转换

中括号内的四个元素就是列表里存储的4个元素。它们的类型分别是“int，string，float，bool”

```python
my_list = [1, 'two', 3.0, True]
```

字符串转列表

```python
list_from_str = list("abc")
```

## 列表的核心操作方法(列表在读取元素的时候，会考虑其类型)

末尾追加元素

```python
my_list.append('new')
```

指定索引插入元素(被插入的元素将会在规定的索引位置出现，而原本的将会向后顺延一个)

```python
my_list.insert(1, 'inserted')
```

连接列表(将extend中的列表接在原本的列表后面)

```python
my_list.extend(["extended", 0.1])
```

移除并返回最后一个元素(也可以不保留被移除的元素)

```python
pop_result = my_list.pop()
```

移除第一个匹配项

```python
my_list.remove('two')
```

替换某个索引对应的元素

```python
my_list[0] = 100
```

查找元素首次出现的索引

```python
list_index = my_list.index(3.0)
```

统计元素出现次数

```python
list_count = my_list.count(1)
```

获取列表长度

```python
list_len = len(my_list)
```

## 列表排序与推导式

```python
list_disorderd = [1, 15, 2, 4, 9, 6]
```

按照升序排列

```python
list_disorderd.sort()
```

借助循环生成函数

* 组成部分[<表达式> for <循环变量> in <循环可迭代对象>]

```python
list_for_loop = [i**2 for i in range(5)]
```

嵌套列表推导式

```python
another_list_for_loop = [[j for j in range(i)] for i in range(10)]
```

生成重复元素列表

```python
list_for_loop_3 = [3 for i in range(10)]
```

# 元组（Tuple）- 有序、不可变、支持混合类型元素

## 元组定义与类型转换

中括号内的三个元素就是元组里存储的3个元素。它们的类型分别是“int，string，float”

```python
my_tuple = (1, 'two', 3.0)
```

单元素元组必须带逗号

```python
single_element_tuple = (42,)
```

列表转元组

```python
tuple_from_list = tuple([5, 6, 7])
```

元组转列表

```python
list_from_tuple = list((4, 5, 6))
```

## 元组的核心操作方法(元组在读取元素的时候，会考虑其类型)

元组拼接（生成新元组，原元组不变）

```python
tuple_concat = my_tuple + (4,)
```

统计元素出现次数

```python
tuple_count = my_tuple.count('two')
```

查找元素首次出现的索引

```python
tuple_index = my_tuple.index(3.0)
```

获取元组长度

```python
tuple_len = len(my_tuple)
```

索引访问元素（仅可读）

```python
tuple_index_access = my_tuple[0]
```

# 集合（Set）- 无序、可变、无重复、元素必须不可变

## 集合定义与类型转换

```python
my_set = {1, 2, 'three', 4}
```

列表转集合（自动去重）

```python
set_from_list = set([4, 5, 5, 6])
```

创建空集合（{}创建空字典）

```python
empty_set = set()
```

## 集合的基础操作方法

添加单个元素

```python
my_set.add(5)
```

移除元素（元素不存在会报错）

```python
my_set.remove(4)
```

移除元素

```python
my_set.discard(1)
```

随机移除并返回一个元素

```python
if my_set:  # 避免空集合pop报错
    set_pop = my_set.pop()
```

获取集合长度

```python
set_len = len(my_set)
```

清空集合

```python
my_set.clear()
```

## 集合的关系运算

```python
set_A = {1, 2, 3}
set_B = {3, 4, 5}
```

并集

```python
set_union = set_A.union(set_B)
```

交集

```python
set_intersection = set_A.intersection(set_B)
```

差集（A有B无）

```python
set_difference = set_A.difference(set_B)
```

判断是否为子集

```python
set_issubset = set_A.issubset({1, 2, 3, 4})
```

# 字典（Dictionary）- 键值对映射、可变、键必须不可变

字典定义与类型转换

```python
my_dict = {'name': 'Alice', 'age': 25, 'city': 'New York'}
```

元组列表转字典

```python
dict_from_pairs = dict([('a', 1), ('b', 2)])
```

关键字参数创建字典

```python
dict_from_kwargs = dict(fruit='apple', color='red')
```

创建空字典

```python
empty_dict = {}
```

## 字典的访问与查询

键访问值（键不存在报错）

```python
dict_value = my_dict['name']
```

安全访问（默认值兜底）

```python
dict_safe_get = my_dict.get('country', 'USA')
```

获取所有键

```python
dict_keys = my_dict.keys()
```

获取所有值

```python
dict_values = my_dict.values()
```

获取所有键值对

```python
dict_items = my_dict.items()
```

获取字典长度（键值对个数）

```python
dict_len = len(my_dict)
```

字典的修改、添加与移除

修改现有键的值

```python
my_dict['age'] = 26
```

添加新键值对

```python
my_dict['email'] = 'alice@example.com'
```

移除并返回指定键的值

```python
dict_pop_value = my_dict.pop('city')
```

合并/更新字典（存在则改，不存在则加）

```python
my_dict.update({'age': 27, 'zip': 10001})
```

清空字典

```python
my_dict.clear()
```
